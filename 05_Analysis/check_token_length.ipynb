```python
import pandas as pd
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained ("aisingapore/Llama-SEA-LION-v3.5-8B-R")

def build_pa_prompt (row) :
  description_detailed = row["targeted_response" ]
  user_message = row["question"]
  context = f"Ini adalah jawaban anda tentang deskripsi \
          \"{description_detailed}\". Saya adalah pengunjung pada Program Studi Informatika kampus UHN I Gusti Bagus Sugriwa \
          ini adalah pertanyaan saya: \"{user_message}\"? Buatlah pendapat yang singkat, sopan dan bersahabat \
          untuk saya dengan gaya bahasa berusia 18 tahun dan jangan memberikan pertanyaan lanjutan. Jawablah dalam 50 kata."
  return context

def time(row):
  return int((row["time_response"].split())[0])
```

```python
file_names = ["top_k_1_in scope.xlsx", "top_k_3_in Scope.xlsx", "top_k_5_in Scope.xlsx"]
for file_name in file_names :
  df = pd.read_excel("./03_Data/"+file_name)
  print (file_name)
  # df["prompt_tokens"] = df["question"].notna() & df[ "targeted_response"].notna()
  df = df[df["Keterangan"] != "Tidak digunakan"]
  df["n_tokens"] = df.apply(lambda r: len(tokenizer.encode(build_pa_prompt(r))), axis=1)
  df["time"] = df.apply(lambda r: time(r), axis=1)
  print (f"Top PA Token: mean={df['n_tokens'].mean():.1f} tokens, SD={df['n_tokens'].std():.1f} tokens, "
        f"median={df['n_tokens'].median():.0f} tokens, n={len(df)}")
  print (f"Top PA Time: mean={df['time'].mean():.3f} seconds, SD={df['time'].std():.3f} seconds, "
        f"median={df['time'].median():.0f} seconds, n={len(df)}")
  print ()
```
